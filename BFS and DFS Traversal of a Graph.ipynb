```python
from collections import deque

# Taking number of nodes
n = int(input("Enter number of nodes: "))

# Taking node names
nodes = []
for i in range(n):
    node = input(f"Enter node {i + 1}: ")
    nodes.append(node)

# Creating adjacency list
graph = {}

for node in nodes:
    graph[node] = []

# Taking edges
e = int(input("Enter number of edges: "))

for i in range(e):
    u, v = input(f"Enter edge {i + 1} (u v): ").split()

    graph[u].append(v)
    graph[v].append(u)   # Remove this line for a directed graph


# ---------------- BFS ----------------
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    order = []

    visited.add(start)

    while queue:
        node = queue.popleft()
        order.append(node)

        for neighbour in graph[node]:
            if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighbour)

    return order


# ---------------- DFS ----------------
def dfs(graph, start):
    visited = set()
    order = []

    def dfs_visit(node):
        visited.add(node)
        order.append(node)

        for neighbour in graph[node]:
            if neighbour not in visited:
                dfs_visit(neighbour)

    dfs_visit(start)

    return order


# Taking starting node
start = input("Enter starting node: ")

# Performing traversals
bfs_order = bfs(graph, start)
dfs_order = dfs(graph, start)

print("\nAdjacency List:")
for node in graph:
    print(node, "->", graph[node])

print("\nBFS Traversal:")
print(" -> ".join(bfs_order))

print("\nDFS Traversal:")
print(" -> ".join(dfs_order))


# ---------------- Tree Diagrams ----------------
print("\nBFS Tree Diagram:")
for i, node in enumerate(bfs_order):
    print("  " * i + node)

print("\nDFS Tree Diagram:")
for i, node in enumerate(dfs_order):
    print("  " * i + node)
```
