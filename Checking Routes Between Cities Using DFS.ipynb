```python
# Taking number of cities
n = int(input("Enter number of cities: "))

cities = []

for i in range(n):
    city = input(f"Enter city {i + 1}: ")
    cities.append(city)

# Creating adjacency list
graph = {}

for city in cities:
    graph[city] = []

# Taking number of roads
e = int(input("Enter number of roads: "))

for i in range(e):
    city1, city2 = input(
        f"Enter road {i + 1} (city1 city2): "
    ).split()

    graph[city1].append(city2)
    graph[city2].append(city1)


# ------------------------------------------------
# DFS to check whether a route exists
# ------------------------------------------------

def route_exists(graph, source, destination):
    visited = set()

    def dfs(city):
        if city == destination:
            return True

        visited.add(city)

        for neighbour in graph[city]:
            if neighbour not in visited:
                if dfs(neighbour):
                    return True

        return False

    return dfs(source)


# ------------------------------------------------
# DFS to find connected components
# ------------------------------------------------

def connected_components(graph):
    visited = set()
    components = []

    for city in graph:

        if city not in visited:
            component = []

            def dfs(city):
                visited.add(city)
                component.append(city)

                for neighbour in graph[city]:
                    if neighbour not in visited:
                        dfs(neighbour)

            dfs(city)
            components.append(component)

    return components


# Display adjacency list
print("\nRoad Network:")
for city in graph:
    print(city, "->", graph[city])


# ------------------------------------------------
# Check route
# ------------------------------------------------

source = input("\nEnter source city: ")
destination = input("Enter destination city: ")

if route_exists(graph, source, destination):
    print(f"\nRoute exists from {source} to {destination}.")
else:
    print(f"\nNo route exists from {source} to {destination}.")


# ------------------------------------------------
# Find connected components
# ------------------------------------------------

components = connected_components(graph)

print("\nConnected Components:")

for i, component in enumerate(components, 1):
    print(f"Component {i}: {component}")
```
