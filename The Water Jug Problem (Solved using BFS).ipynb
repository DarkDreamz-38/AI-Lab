```python
from collections import deque

# Fixed values from the problem
capacity_a = 4
capacity_b = 3
target = 2


# Generate all possible operations
def get_next_states(a, b):

    states = []

    # Fill Jug A
    states.append(((4, b), "Fill Jug A"))

    # Fill Jug B
    states.append(((a, 3), "Fill Jug B"))

    # Empty Jug A
    states.append(((0, b), "Empty Jug A"))

    # Empty Jug B
    states.append(((a, 0), "Empty Jug B"))

    # Pour Jug A -> Jug B
    amount = min(a, 3 - b)
    states.append(
        ((a - amount, b + amount), "Pour Jug A -> Jug B")
    )

    # Pour Jug B -> Jug A
    amount = min(b, 4 - a)
    states.append(
        ((a + amount, b - amount), "Pour Jug B -> Jug A")
    )

    return states


# BFS function
def water_jug_bfs():

    queue = deque()

    visited = set()

    # Starting state
    queue.append(((0, 0), []))

    visited.add((0, 0))

    while queue:

        (a, b), path = queue.popleft()

        # Check target
        if a == target or b == target:
            return path, (a, b)

        # Generate next states
        for state, operation in get_next_states(a, b):

            if state not in visited:

                visited.add(state)

                queue.append(
                    (state, path + [(state, operation)])
                )

    return None, None


# Solve the problem
solution, target_state = water_jug_bfs()


# Display result
if solution:

    print("\nMinimum sequence of steps:\n")

    print("Initial State: (0, 0)\n")

    for i, (state, operation) in enumerate(solution, 1):

        print(f"Step {i}: {operation}")
        print(f"        State: {state}\n")

    print("Target reached!")
    print(f"State: {target_state}")

else:

    print("No solution exists.")
```
