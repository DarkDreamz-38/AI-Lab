```python
from collections import deque

# Taking user input
capacity_a = int(input("Enter capacity of Jug A: "))
capacity_b = int(input("Enter capacity of Jug B: "))
target = int(input("Enter target amount: "))


def get_next_states(a, b):
    states = []

    # Fill Jug A
    states.append(((capacity_a, b), "Fill Jug A"))

    # Fill Jug B
    states.append(((a, capacity_b), "Fill Jug B"))

    # Empty Jug A
    states.append(((0, b), "Empty Jug A"))

    # Empty Jug B
    states.append(((a, 0), "Empty Jug B"))

    # Pour Jug A -> Jug B
    amount = min(a, capacity_b - b)
    states.append(
        ((a - amount, b + amount), "Pour Jug A -> Jug B")
    )

    # Pour Jug B -> Jug A
    amount = min(b, capacity_a - a)
    states.append(
        ((a + amount, b - amount), "Pour Jug B -> Jug A")
    )

    return states


def water_jug_bfs():
    queue = deque()
    visited = set()

    # Starting state
    start = (0, 0)

    queue.append((start, []))
    visited.add(start)

    while queue:
        (a, b), path = queue.popleft()

        # Check whether target has been reached
        if a == target or b == target:
            return path + [((a, b), "Target reached")]

        # Generate all possible next states
        for new_state, operation in get_next_states(a, b):

            if new_state not in visited:
                visited.add(new_state)

                queue.append(
                    (new_state, path + [(new_state, operation)])
                )

    return None


solution = water_jug_bfs()

if solution:
    print("\nMinimum sequence of steps:\n")

    print("Initial State: (0, 0)")

    for i, (state, operation) in enumerate(solution, 1):
        print(f"Step {i}: {operation}")
        print(f"        State: {state}")

else:
    print("No solution exists.")
```
